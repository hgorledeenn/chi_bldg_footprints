File from nb8

```python
import geopandas as gpd
from plotnine import *
import matplotlib.pyplot as plt
```

```python
df = gpd.read_file("data/nb8_buildings_coded.geojson")
```

```python
df.head()
```

```python
df[df['noise_from_one_train']>0].shape
```

```python
fig, ax = plt.subplots(1, 1, figsize=(50, 50))

# 3. Plot the map, filling polygons based on your target column
# 'column' specifies the data variable to use for colors
# 'cmap' sets the color scheme (e.g., 'viridis', 'plasma', 'Blues')
df.plot(
    column='daily_noise', 
    cmap='viridis', 
    legend=True, 
    ax=ax,
    edgecolor='black', 
    linewidth=0.1
)
```

```python
plot = (
    ggplot(df) +
    aes(x='noise_from_one_train') +
    geom_histogram(binwidth=10)
)
plot
```

```python
plot = (
    ggplot(df) +
    aes(y='factor(num_radials)', x='daily_noise') +
    geom_point(alpha=0.01)
)
plot
```

```python
plot = (
    ggplot(df) +
    aes(x='daily_noise', y='intersected_daily_noise') +
    geom_point(alpha=0.5)
)
plot
```
